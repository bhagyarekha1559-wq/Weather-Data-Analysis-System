```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/bhagyarekha1559-wq/Weather-Data-Analysis-System/main/climate_code.csv"

df = pd.read_csv(url)

print(df.head())

print("=" * 60)
print("           WEATHER DATA ANALYSIS SYSTEM")
print("=" * 60)

print("\nDATASET LOADED SUCCESSFULLY!")

print("\nFirst 5 Records:")
print(df.head())

print("\n" + "=" * 60)
print("DATASET EXPLORATION")
print("=" * 60)
print("\nDataset Shape:")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
print("\nColumn Names:")
print(list(df.columns))
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isnull().sum())
print("\nStatistical Summary:")
print(df.describe())
print("\n" + "=" * 60)
print("DATA PREPROCESSING")
print("=" * 60)
df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)

df["Temperature"] = pd.to_numeric(
    df["Temperature"],
    errors="coerce"
)
df["Humidity"] = pd.to_numeric(
    df["Humidity"],
    errors="coerce"
)
print("\nDate and numeric columns processed successfully.")
print("\n" + "=" * 60)
print("DATA CLEANING")
print("=" * 60)
missing_before = df.isnull().sum().sum()
print("\nTotal missing values before cleaning:",
      missing_before)
df = df.dropna()
duplicates_removed = df.duplicated().sum()
df = df.drop_duplicates()
df = df.sort_values("Date")
print("Duplicate records removed:",
      duplicates_removed)
print("Total missing values after cleaning:",
      df.isnull().sum().sum())
print("\nCleaned Dataset:")
print(df)
cleaned_file = "cleaned_climate_codecsv"
df.to_csv(
    cleaned_file,
    index=False
)
print("\nCleaned dataset saved as:",
      cleaned_file)
print("\n" + "=" * 60)
print("TEMPERATURE ANALYSIS")
print("=" * 60)
minimum_temperature = df["Temperature"].min()
maximum_temperature = df["Temperature"].max()
average_temperature = df["Temperature"].mean()
print(
    "\nMinimum Temperature :",
    round(minimum_temperature, 2),
    "°C"
)
print(
    "Maximum Temperature :",
    round(maximum_temperature, 2),
    "°C"
)
print(
    "Average Temperature :",
    round(average_temperature, 2),
    "°C"
)
print("\n" + "=" * 60)
print("HUMIDITY ANALYSIS")
print("=" * 60)
average_humidity = df["Humidity"].mean()
minimum_humidity = df["Humidity"].min()
maximum_humidity = df["Humidity"].max()
print(
    "\nMinimum Humidity :",
    round(minimum_humidity, 2),
    "%"
)
print(
    "Maximum Humidity :",
    round(maximum_humidity, 2),
    "%"
)
print(
    "Average Humidity :",
    round(average_humidity, 2),
    "%"
)
print("\n" + "=" * 60)
print("WEATHER CONDITION ANALYSIS")
print("=" * 60)
weather_counts = df["Weather"].value_counts()
print("\nWeather Conditions:")
print(weather_counts)
most_common_weather = weather_counts.idxmax()
print(
    "\nMost Common Weather Condition:",
    most_common_weather
)
print("\n" + "=" * 60)
print("RUNTIME TEMPERATURE CONVERSION")
print("=" * 60)
print("\n1. Celsius to Fahrenheit")
print("2. Fahrenheit to Celsius")
choice = input("\nEnter your choice (1 or 2): ")
runtime_temperature = None
runtime_celsius = None
runtime_fahrenheit = None
if choice == "1":
    runtime_celsius = float(
        input("Enter temperature in Celsius (°C): ")
    )
    runtime_fahrenheit = (
        runtime_celsius * 9 / 5
    ) + 32
    print("\nCONVERSION RESULT")
    print("-----------------")
    print(
        f"{runtime_celsius:.2f} °C = "
        f"{runtime_fahrenheit:.2f} °F"
    )
elif choice == "2":
    runtime_fahrenheit = float(
        input("Enter temperature in Fahrenheit (°F): ")
    )
    runtime_celsius = (
        runtime_fahrenheit - 32
    ) * 5 / 9
    print("\nCONVERSION RESULT")
    print("-----------------")
    print(
        f"{runtime_fahrenheit:.2f} °F = "
        f"{runtime_celsius:.2f} °C"
    )
else:
    print("\nInvalid choice!")

if choice in ("1", "2"):
    print("\n" + "=" * 60)
    print("YOUR RUNTIME CONVERSION RESULT")
    print("=" * 60)
    print(f"Temperature in Celsius: {runtime_celsius:.2f} °C")
    print(f"Temperature in Fahrenheit: {runtime_fahrenheit:.2f} °F")
    print("=" * 60)

input("\nPress Enter to continue to dataset conversion...")

print("\n" + "=" * 60)
print("DATASET TEMPERATURE CONVERSION")
print("=" * 60)
df["Temperature_F"] = (
    df["Temperature"] * 9 / 5
) + 32
print("\nActual Dataset:")
print(
    df[
        ["Date", "Temperature", "Temperature_F"]
    ].to_string(index=False)
)
print("\n" + "=" * 60)
print("TEMPERATURE LINE GRAPH")
print("=" * 60)
print("\nGenerating line graph using actual dataset...")
plt.figure(figsize=(12,7))
plt.plot(
    df["Date"],
    df["Temperature"],
    marker="o",
    linewidth=1,
    label="Temperature (°C)"
)
if runtime_celsius is not None:
    plt.axhline(
        y=runtime_celsius,
        linestyle="--",
        label=f"Runtime Input: {runtime_celsius:.2f} °C"
    )
plt.title(
    "Temperature Trend Over Time",
    fontsize=18
)
plt.xlabel(
    "Date",
    fontsize=14
)
plt.ylabel(
    "Temperature (°C)",
    fontsize=14
)
plt.xticks(
    rotation=45,
    fontsize=10
)
plt.yticks(
    fontsize=10
)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(
    "temperature_trend.png",
    dpi=300
)
plt.show()
print("\n" + "=" * 60)
print("TEMPERATURE HISTOGRAM")
print("=" * 60)
print("\nGenerating histogram using actual dataset...")
plt.figure(figsize=(10,6))
plt.hist(
    df["Temperature"],
    bins=10,
    edgecolor="black"
)
if runtime_celsius is not None:
    plt.axvline(
        x=runtime_celsius,
        linestyle="--",
        linewidth=2,
        label=f"Runtime Input: {runtime_celsius:.2f} °C"
    )
plt.title(
    "Temperature Distribution",
    fontsize=18
)
plt.xlabel(
    "Temperature (°C)",
    fontsize=14
)
plt.ylabel(
    "Frequency",
    fontsize=14
)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(
    "temperature_distribution.png",
    dpi=300
)
plt.show()
print("\n" + "=" * 60)
print("GENERATED GRAPHS")
print("=" * 60)
print("\n1. Temperature Trend Line Graph")
print("   → Based on actual Date and Temperature data")
print("\n2. Temperature Distribution Histogram")
print("   → Based on actual Temperature values")
print("\nGraphs generated successfully!")
print("\n" + "=" * 60)
print("HUMIDITY TREND OVER TIME")
print("=" * 60)
print("\nGenerating humidity trend graph...")
plt.figure(figsize=(12,7))
plt.plot(
    df["Date"],
    df["Humidity"],
    marker="o",
    linewidth=1,
    label="Humidity (%)"
)
plt.title(
    "Humidity Trend Over Time",
    fontsize=18
)
plt.xlabel(
    "Date",
    fontsize=14
)
plt.ylabel(
    "Humidity (%)",
    fontsize=14
)
plt.xticks(
    rotation=45,
    fontsize=10
)
plt.yticks(
    fontsize=10
)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig(
    "humidity_trend.png",
    dpi=300
)
plt.show()
print("\n" + "=" * 60)
print("TEMPERATURE VS HUMIDITY")
print("=" * 60)
print("\nGenerating scatter plot...")
plt.figure(figsize=(10,6))
plt.scatter(
    df["Temperature"],
    df["Humidity"],
    alpha=0.6
)
plt.title(
    "Temperature vs Humidity",
    fontsize=18
)
plt.xlabel(
    "Temperature (°C)",
    fontsize=14
)
plt.ylabel(
    "Humidity (%)",
    fontsize=14
)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.grid(True)
plt.tight_layout()
plt.savefig(
    "temperature_vs_humidity.png",
    dpi=300
)
plt.show()
print("\n" + "=" * 60)
print("WEATHER CONDITION FREQUENCY")
print("=" * 60)
print("\nGenerating weather condition frequency graph...")
weather_counts = df["Weather"].value_counts()
plt.figure(figsize=(10,6))
plt.bar(
    weather_counts.index,
    weather_counts.values
)
plt.title(
    "Weather Condition Frequency",
    fontsize=18
)
plt.xlabel(
    "Weather Condition",
    fontsize=14
)
plt.ylabel(
    "Number of Records",
    fontsize=14
)
plt.xticks(
    rotation=45,
    fontsize=10
)
plt.yticks(
    fontsize=10
)
plt.grid(axis="y")
plt.tight_layout()
plt.savefig(
    "weather_condition_frequency.png",
    dpi=300
)
plt.show()
print("\n" + "=" * 60)
print("AVERAGE TEMPERATURE BY WEATHER CONDITION")
print("=" * 60)
print("\nGenerating average temperature graph...")
average_temp_weather = (
    df.groupby("Weather")["Temperature"].mean()
)
plt.figure(figsize=(10,6))
plt.bar(
    average_temp_weather.index,
    average_temp_weather.values
)
plt.title(
    "Average Temperature by Weather Condition",
    fontsize=18
)
plt.xlabel(
    "Weather Condition",
    fontsize=14
)
plt.ylabel(
    "Average Temperature (°C)",
    fontsize=14
)
plt.xticks(
    rotation=45,
    fontsize=10
)
plt.yticks(
    fontsize=10
)
plt.grid(axis="y")
plt.tight_layout()
plt.savefig(
    "average_temperature_by_weather.png",
    dpi=300
)
plt.show()
print("\n" + "=" * 60)
print("MONTHLY AVERAGE TEMPERATURE")
print("=" * 60)
print("\nGenerating monthly average temperature graph...")
monthly_temperature = (
    df.groupby(
        df["Date"].dt.month
    )["Temperature"].mean()
)
plt.figure(figsize=(10,6))
plt.plot(
    monthly_temperature.index,
    monthly_temperature.values,
    marker="o",
    linewidth=2
)
plt.title(
    "Monthly Average Temperature",
    fontsize=18
)
plt.xlabel(
    "Month",
    fontsize=14
)
plt.ylabel(
    "Average Temperature (°C)",
    fontsize=14
)
plt.xticks(
    range(1, 13),
    [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug",
        "Sep", "Oct", "Nov", "Dec"
    ],
    fontsize=10
)
plt.yticks(
    fontsize=10
)
plt.grid(True)
plt.tight_layout()
plt.savefig(
    "monthly_average_temperature.png",
    dpi=300
)
plt.show()
print("\n" + "=" * 60)
print("ALL GENERATED GRAPHS")
print("=" * 60)
print("\n1. Temperature Trend Line Graph")
print("   → Based on actual Date and Temperature data")
print("\n2. Temperature Distribution Histogram")
print("   → Based on actual Temperature values")
print("\n3. Humidity Trend Over Time")
print("   → Shows humidity variation across dates")
print("\n4. Temperature vs Humidity Scatter Plot")
print("   → Shows relationship between temperature and humidity")
print("\n5. Weather Condition Frequency")
print("   → Shows number of records for each weather condition")
print("\n6. Average Temperature by Weather Condition")
print("   → Shows average temperature for each weather condition")
print("\n7. Monthly Average Temperature")
print("   → Shows average temperature for each month")
print("\nAll graphs generated successfully!")
```
